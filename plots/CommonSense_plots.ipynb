```python
import numpy as np
import matplotlib.pyplot as plt

q = np.array((10,10)).reshape(1,2)

r = 16
nu=0.5
s = 1/(r**nu)
x = []
y = []
for i in range(10000):
    A = s* np.random.randn(r, 2)/r
    z = np.matmul(np.matmul(q, A.T), A)
    x.append(z[0,0])
    y.append(z[0,1])

plt.scatter(x,y, s=1, alpha=1, color='#A0C4FF')
p = (2*nu+1)
y = q/(r**p)
plt.plot(y[0,0], y[0,1],'o', color='#0047AB', markersize=5, label=r'$\mathbf{q}$/256  (r=16)')





r = 32
nu=0.5
s = 1/(r**nu)
x = []
y = []
for i in range(5000):
    A =  np.random.randn(r, 2)/r
    z = np.matmul(np.matmul(q, s * A.T), s * A)
    x.append(z[0,0])
    y.append(z[0,1])

plt.scatter(x, y, s=1, alpha=1, color='#FFCAAF')
p = 2*nu+1
y = q/(r**p)
plt.plot(y[0,0], y[0,1],'s', color='#D35400', markersize=5, label=r'$\mathbf{q}$/1024  (r=32)')


plt.axhline(0, color='black', linewidth=1) # bold horizontal line at y=0
plt.axvline(0, color='black', linewidth=1)

title_str = r"Scatter plot of $\mathbf{q}(s^2\mathbf{A}_\ell^{0\top}\mathbf{A}_\ell^0)$ for $\mathbf{q}=(10,10), \nu=0.5$ (rs-LoRA)"
plt.title(title_str)
plt.xlabel('X coordinate')
plt.ylabel('Y coordinate')
plt.legend(loc='upper right')
plt.xlim([-0.01, 0.1])
plt.ylim([-0.01, 0.1])
plt.grid(True, linestyle='--', alpha=0.8)
plt.gca().set_aspect('equal')
plt.tight_layout()
plt.savefig("scatterplot.png", dpi=300, bbox_inches='tight')
```

```python
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np

# Data Definitions

x4 = [2, 3, 4, 5, 6, 7, 8]
y4 = [68.82, 69.31, 69.91, 72.85, 70.89, 65.18, 34.54]

x64 = [1, 2, 3, 4, 5, 6, 7]
y64 = [77.85, 78.15, 78.66, 78.84, 76.73, 68.28, 34.54]

datasets = [
    (x64, y64, "r=64"),
    (x4, y4, "r=4")
]

# Define unique markers for each line
markers = ['s', 'p']

# Learning Rate Labels
lr_labels = ["4e-4", "6e-4", "8e-4", "1e-3", "2e-3", "4e-3", "6e-3", "6e-3"]
x_ticks = np.arange(1, 9, 1)

plt.figure(figsize=(10, 5))

# Widened linspace (0.0 to 1.05) for better color separation at the dark end
colors = cm.viridis(np.linspace(0.0, 1.05, len(datasets)))

for i, (x, y, label) in enumerate(datasets):
    plt.plot(x, y, 
             marker=markers[i], # Unique marker per dataset
             markerfacecolor=colors[i], 
             markeredgecolor='white', 
             markeredgewidth=0.7,
             linestyle='-', 
             color=colors[i], 
             label=label, 
             linewidth=2.2, 
             markersize=10)

# Axis Formatting
plt.xlabel("AdamW learning rate", fontsize=16)
plt.ylabel("average test accuracy", fontsize=16)
plt.title(r"Avg test accuracy on Commonsense Reasoning vs. AdamW learning rate (trainable $\mathbf{A}$ and $\mathbf{B}$)", fontsize=16, pad=15)

# Map the numbers 1-37 to the LR strings (default size 8)
plt.xticks(x_ticks, lr_labels, rotation=45, fontsize=8)
plt.ylim ([60, 80])

# Customize color and font size for specific tick labels
target_blue_lrs = ["1e-3", "2e-3"]
for tick in plt.gca().get_xticklabels():
    if tick.get_text() in target_blue_lrs:
        if tick.get_text() == "1e-3":
            tick.set_color(colors[0])
            tick.set_fontsize(8)
        elif tick.get_text() == "2e-3":
            tick.set_color(colors[1])
            tick.set_fontsize(8)

plt.grid(True, which='both', linestyle='--', alpha=1.0)
plt.legend(frameon=True, loc='upper right', fontsize=15)

plt.tight_layout()
plt.savefig("rank_lr_trainA.png", dpi=300, bbox_inches='tight')
plt.show()
```
